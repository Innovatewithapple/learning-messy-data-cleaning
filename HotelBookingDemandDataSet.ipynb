```python
import pandas as pd
import numpy as np

df_raw = pd.read_csv("InputFiles/hotel_bookings.csv")
df_raw.head()
```

```python
df_Copy = df_raw.copy()
df_Copy.info()
```

```python
df_Copy.columns
```

```python
df_Copy["reservation_status_date"] = pd.to_datetime(df_Copy["reservation_status_date"])
df_Copy = df_Copy.sort_values('reservation_status_date').reset_index(drop=True)
df_Copy.head()
```

```python
# types_of_status = df_Copy["reservation_status"].unique()
reservation_status_withoutCanceled = df_Copy[df_Copy["reservation_status"] == "Check-Out"].copy()
print(f"Total rows: {len(reservation_status_withoutCanceled)}")
print(f"Shape: {reservation_status_withoutCanceled.shape}")
print(f"Rows: {reservation_status_withoutCanceled.shape[0]}")
print(f"Columns: {reservation_status_withoutCanceled.shape[1]}")
print(f"Total elements: {reservation_status_withoutCanceled.size}")
```

```python
reservation_status_withoutCanceled["Total_nights"] = reservation_status_withoutCanceled["stays_in_weekend_nights"] + reservation_status_withoutCanceled["stays_in_week_nights"]
reservation_status_withoutCanceled.head()
```

```python
month_stays = (reservation_status_withoutCanceled
               .groupby("arrival_date_month")
               .agg(Total_stays=("arrival_date_month","size"),
                    Total_nights=("Total_nights",sum),
                    average_price=("adr","mean"))
               .reset_index())
month_stays["average_price"] = month_stays["average_price"].round(2)
```

```python
month_stays.head(12)
```

```python
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December"]
month_stays["arrival_date_month"] = pd.Categorical(month_stays["arrival_date_month"],categories=month_order,ordered=True)
month_stays = month_stays.sort_values("arrival_date_month")
```
